```python
import random
import pandas as pd
from stxntx.config import RAW_DIR
from stxntx.sources import prepare_season, make_client
from stxntx.possessions import possession_table
from stxntx.tfo_flags import flag_period_ends, baseline_possessions

prepare_season(2016)
ids = sorted(p.stem[6:] for p in (RAW_DIR / "pbp").glob("stats_002*.json"))
sample = random.Random(7).sample(ids, 250)

client = make_client(2016)
tables, failed = [], []
for gid in sample:
    try:
        tables.append(possession_table(client.Game(gid), gid))
    except Exception as e:
        failed.append((gid, type(e).__name__))

poss = pd.concat(tables)
flags = flag_period_ends(poss)
print(len(tables), "games parsed,", len(failed), "failed")
flags[["tfo_attempted", "tfo_achieved"]].value_counts()
```
